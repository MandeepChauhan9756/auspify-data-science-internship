```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)
```

```python
df = pd.read_csv("Dataset.csv")
df.head()
```

```python
print("Daraset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()
print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

missing_percentage
```

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print("Shape after removing duplicates:", df.shape)
```

```python
text_columns = df.select_dtypes(include="str").columns

print("Text columns:")
print(text_columns.tolist())

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

print(df["date_added"].dtype)
```

```python
df["added_year"] = df["date_added"].dt.year
df["added_month"] = df["date_added"].dt.month
df["added_day"] = df["date_added"].dt.day

df[["date_added", "added_year", "added_month", "added_day"]].head()
```

```python
df["duration_value"] = pd.to_numeric(
    df["duration"].str.extract(r"(\d+)")[0],
    errors="coerce"
)

df["duration_type"] = df["duration"].str.extract(
    r"(min|Seasons?)",
    expand=False
)
```

```python
df[[
    "duration", "duration_value", "duration_type"
]].head(10)
```

```python
print("Final Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
